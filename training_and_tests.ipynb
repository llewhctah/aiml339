```python
# Load model 1
baseline_best_model = YOLO("/content/drive/MyDrive/SONICNN/results/hand_detection/egohands/egohands_model/weights/best.pt")
```

### **- Test Model 1 on EgoHands test set**


```python
egohands_test_metrics = baseline_best_model.val(
    data="/content/egohands/data.yaml",
    split="test",
    imgsz=640,
    conf=0.001,
    iou=0.7,
    plots=True,
    exist_ok=True,
    project="/content/drive/MyDrive/SONICNN/results/hand_detection/egohands/tests",
    name="test_on_egohands",
)
```

### **- Test Model 1 on Webcam test set**


```python
webcam_test_metrics = baseline_best_model.val(
    data="/content/webcam/webcam_dataset/data.yaml",
    split="test",
    imgsz=640,
    conf=0.001,
    iou=0.7,
    plots=True,
    exist_ok=True,
    project="/content/drive/MyDrive/SONICNN/results/hand_detection/egohands/tests",
    name="test_on_webcam",
)
```

**Access results**


```python
def printresults(test_metrics, model_name, test_name):
  print(f"{model_name} — {test_name} test set results")
  print(f"  Precision:  {test_metrics.box.mp:.3f}")
  print(f"  Recall:     {test_metrics.box.mr:.3f}")
  print(f"  mAP50:      {test_metrics.box.map50:.3f}")
  print(f"  mAP50-95:   {test_metrics.box.map:.3f}")
```

```python
# Print Egohands test results
printresults(egohands_test_metrics, "Model 1", "EgoHands")
```

```python
# Print Webcam test results
printresults(webcam_test_metrics, "Model 1", "Webcam")
```

# **Model 2: Model 1 fine-tuned on webcam data**


```python
for seed in [0, 1, 2, 3, 4]:
    model_ft = YOLO("/content/drive/MyDrive/SONICNN/results/hand_detection/egohands/egohands_model/weights/best.pt")

    model_ft.train(
        data="/content/webcam/webcam_dataset/data.yaml",
        epochs=30,
        imgsz=640,
        batch=16,
        seed=seed,
        patience=10,
        optimizer="AdamW",
        lr0=0.001,
        project="/content/drive/MyDrive/SONICNN/results/hand_detection/webcam",
        name=f"webcam_model_seed{seed}",
    )
```

## **- Test model 2 on Webcam test set**

```python
model2_webcam = []

for seed in [0, 1, 2, 3, 4]:
    model = YOLO(f"/content/drive/MyDrive/SONICNN/results/hand_detection/webcam/webcam_model_seed{seed}/weights/best.pt")

    metrics = model.val(
        data="/content/webcam/webcam_dataset/data.yaml",
        split="test",
        imgsz=640,
        conf=0.001,
        iou=0.7,
        plots=True,
        exist_ok=True,
        project="/content/drive/MyDrive/SONICNN/results/hand_detection/webcam/tests",
        name=f"seed{seed}_test_on_webcam",
    )

    model2_webcam.append(metrics)
    printresults(metrics, f"Model 2 (seed {seed})", "Webcam")
```

## **- Test model 2 on EgoHands test set**

```python
model2_egohands = []
for seed in [0, 1, 2, 3, 4]:
    model = YOLO(f"/content/drive/MyDrive/SONICNN/results/hand_detection/webcam/webcam_model_seed{seed}/weights/best.pt")

    metrics = model.val(
        data="/content/egohands/data.yaml",
        split="test",
        imgsz=640,
        conf=0.001,
        iou=0.7,
        plots=True,
        exist_ok=True,
        project="/content/drive/MyDrive/SONICNN/results/hand_detection/webcam/tests",
        name=f"seed{seed}_test_on_egohands",
    )

    model2_egohands.append(metrics)
    printresults(metrics, f"Model 2 (seed {seed})", "EgoHands")
```

**Access results**


```python
import numpy as np

def print_mean_std(results, model_name, test_name):
    print(f"{model_name} — {test_name} test set, mean ± std over {len(results)} runs")
    print(f"  Precision:  {np.mean([m.box.mp for m in results]):.3f} ± {np.std([m.box.mp for m in results], ddof=1):.3f}")
    print(f"  Recall:     {np.mean([m.box.mr for m in results]):.3f} ± {np.std([m.box.mr for m in results], ddof=1):.3f}")
    print(f"  mAP50:      {np.mean([m.box.map50 for m in results]):.3f} ± {np.std([m.box.map50 for m in results], ddof=1):.3f}")
    print(f"  mAP50-95:   {np.mean([m.box.map for m in results]):.3f} ± {np.std([m.box.map for m in results], ddof=1):.3f}")
```

```python
# Print webcam test result
print_mean_std(model2_webcam, "Model 2", "Webcam")
```

```python
# Print egoHands test result
print_mean_std(model2_egohands, "Model 2", "EgoHands")
```
